1. Write a program that asks the user for a number of days. The program then
prints out the number of seconds in the amount of days given.

```python

days = int(input("Enter the number of days: "))

seconds = days * 24 * 60 * 60
print("The number of seconds is:", seconds)
```

2. Write a program that asks a user to input the radius then the program
calculates the volume of a sphere (the formula for the volume is (4/3)πr3
).
Use the exponential operator in python to compute (r3
).

```python

import math


radius = float(input("Enter the radius of the sphere: "))
volume = (4 / 3) * math.pi * (radius ** 3)


print("The volume of the sphere is:", volume)
```

3. Using functions, write a program to compute the area and perimeter of a
square. The program should ask the user to enter a number corresponding to
the side length of the square and display the area and perimeter of the
square.

```python
def calculate_area(side):
    return side ** 2
def calculate_perimeter(side):
    return 4 * side
side = float(input("Enter the side length of the square: "))
area = calculate_area(side)
perimeter = calculate_perimeter(side)
print("Area of the square:", area)
print("Perimeter of the square:", perimeter)
```

4. Write a program using functions that determines whether a character input
by a user is uppercase or lower case.

```python
def check_case(character):
    if character.isupper():
        return "Uppercase"
    elif character.islower():
        return "Lowercase"
    else:
        return "The character is neither uppercase nor lowercase"
character = input("Enter a character: ")
print(check_case(character))
```

5. The following is pseudocode for a program being designed. Write the
Python program equivalent of the same.
BEGIN
SET x TO 0, y TO 20
REPEAT
SUBTRACT 4 FROM y
ADD 2/y TO x
UNTIL
y IS LESS THAN 6
DISPLAY x
END

```python
x = 0
y = 20
while True:
    # Subtract 4 from y
    y = y - 4

    # Add 2/y to x
    x = x + (2 / y)
    if y < 6:
        break
print("The value of x is:", x)
```

6. Write a Python program that does the following:
• Uses a loop for a user to continually input 5 values to populate an
array.
• Calculates and displays the average of the values input into the array.

```python
values = []
for i in range(5):
    value = float(input(f"Enter value {i + 1}: "))
    values.append(value)
average = sum(values) / len(values)
print("Values entered:", values)
print("Average:", average)
```
